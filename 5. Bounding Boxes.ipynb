```python
import pytesseract
import cv2
```

```python
image = cv2.imread("G:\OCR\index_02.JPG")
base_image = image.copy()
```

## grayscale

```python
gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
cv2.imwrite("temp/index_gray.png",gray)
```

## blur

```python
blur = cv2.GaussianBlur(gray,(7,7),0)
```

```python
cv2.imwrite("temp/index_blur.png",blur)
```

## Thresh

```python
thresh = cv2.threshold(blur,0,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)[1]
```

```python
cv2.imwrite("temp/index_thresh.png",thresh)
```

## Kernel

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT,(3,13))
```

```python
cv2.imwrite("temp/index_kernel.png",kernel)
```

## Dilation

```python
dilate = cv2.dilate(thresh,kernel,iterations=1)
```

```python
cv2.imwrite("temp/index_dilate.png",dilate)
```

## Contours

```python
cnts = cv2.findContours(dilate,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
cnts = cnts[0] if len(cnts) == 2 else cnts[1]
```

```python
cnts = sorted(cnts,key=lambda x:cv2.boundingRect(x)[0])
```

## ROI

```python
for c in cnts:
    x,y,w,h = cv2.boundingRect(c)
    if h > 200 and w> 20 :
        roi = image[y:y+h, x:x+w]
        cv2.imwrite("temp/index_roi.png",roi) 
        cv2.rectangle(image,(x,y),(x+w,y+h),(36,255,12),2)
cv2.imwrite("temp/index_bbbox.png",image)
```

```python
result = []
for c in cnts:
    x,y,w,h = cv2.boundingRect(c)
    if h > 200 and w> 20 :
        roi = image[y:y+h, x:x+w]
        cv2.rectangle(image,(x,y),(x+w,y+h),(36,255,12),2)
        ocr_result = pytesseract.image_to_string(roi)
        ocr_Result = ocr_result.split("\n")
        for item in ocr_result:
            result.append(item)
cv2.imwrite("temp/index_bbox_new.png",image)
```
